```python
import pandas as pd

df = pd.read_csv("pizza_sales.csv")
df = df.drop_duplicates()

# dono formats (/ aur -) ko ek jaisa karke din-pehle format me padho
df["order_date"] = pd.to_datetime(df["order_date"].str.replace("/", "-"), format="%d-%m-%Y")
print("bad dates:", df["order_date"].isna().sum())

for c in ["pizza_id", "order_id", "quantity"]:
    df[c] = df[c].astype(int)

df["month_name"] = df["order_date"].dt.month_name()
df["month_no"] = df["order_date"].dt.month
df["day_name"] = df["order_date"].dt.day_name()
df["day_no"] = df["order_date"].dt.dayofweek + 1
df["order_hour"] = pd.to_datetime(df["order_time"], format="%H:%M:%S").dt.hour

df.to_csv("pizza_clean.csv", index=False)
print(df.shape)
```

```python
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

df = pd.read_csv("pizza_clean.csv", parse_dates=["order_date"])
url = URL.create("postgresql+psycopg2", username="postgres",
                 password="your password", host="localhost", port=5432, database="pizza")
engine = create_engine(url)
df.to_sql("pizza_sales", engine, if_exists="append", index=False)
print("Done", len(df))
```
